```python
# Import libraries for visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Display the exact column names
print(df.columns.tolist())
```

```python
# Count the number of flowers in each species
df["species"].value_counts().plot(kind="bar")

plt.title("Number of Iris Flowers by Species")
plt.xlabel("Species")
plt.ylabel("Count")

plt.show()
```

```python
print(df.columns.tolist())
```

```python
# Display the percentage of each Iris species
df["species"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Iris Species Distribution")
plt.ylabel("")

plt.show()
```

```python
# Distribution of sepal length
plt.hist(df["sepal_length"], bins=10)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")

plt.show()
```

```python
# Distribution of petal length
plt.hist(df["petal_length"], bins=10)

plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Frequency")

plt.show()
```

```python
# Relationship between sepal length and sepal width
plt.scatter(df["sepal_length"], df["sepal_width"])

plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")

plt.show()
```

```python
# Relationship between petal length and petal width
plt.scatter(df["petal_length"], df["petal_width"])

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")

plt.show()
```

```python
# Box plot of numerical features
df.boxplot(
    column=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ]
)

plt.title("Box Plot of Iris Measurements")
plt.ylabel("Measurement")

plt.show()
```

```python
# Calculate correlation between numerical features
numeric_df = df.select_dtypes(include="number")

sns.heatmap(numeric_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Compare petal length for different species
sns.boxplot(x="species", y="petal_length", data=df)

plt.title("Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")

plt.show()
```
